```python
import numpy as np
import cppma
import pandas as pd
import time

# -----------------------------------
# Generate Dummy Prices
# -----------------------------------

N = 1_000_000

prices = np.random.randn(N).cumsum() + 100

# -----------------------------------
# Run C++ Moving Average
# -----------------------------------

start = time.time()

ma_cpp = cppma.moving_average(prices, 50)

end = time.time()

print("C++ MA:")
print(ma_cpp[:10])

print(f"\nC++ Execution Time: {end - start:.6f} seconds")


# -----------------------------------
# Compare with Pandas
# -----------------------------------

start = time.time()

ma_pd = (
    pd.Series(prices)
    .rolling(50)
    .mean()
    .fillna(0)
    .values
)

end = time.time()

print("\nPandas MA:")
print(ma_pd[:10])

print(f"\nPandas Execution Time: {end - start:.6f} seconds")


# -----------------------------------
# Verify Numerical Equality
# -----------------------------------

print(
    "\nOutputs Match:",
    np.allclose(ma_cpp, ma_pd)
)
```
