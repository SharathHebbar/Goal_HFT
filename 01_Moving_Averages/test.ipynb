```python
import numpy as np
import cppma
import pandas as pd
import time

# -----------------------------------
# Generate Dummy Prices
# -----------------------------------

N = 10_000_000
window_size = 50
prices = np.random.randn(N).cumsum() + 1000

# -----------------------------------
# Run C++ Moving Average
# -----------------------------------

start = time.time()

ma_cpp = cppma.moving_average(prices, window_size)

end = time.time()

print("C++ MA:")
print(ma_cpp[window_size: window_size + 10])

print(f"\nC++ Execution Time: {end - start:.6f} seconds")

cpp_time = end - start

# -----------------------------------
# Compare with Pandas
# -----------------------------------

start = time.time()

ma_pd = (
    pd.Series(prices)
    .rolling(window_size)
    .mean()
    .fillna(0)
    .values
)

end = time.time()

print("\nPandas MA:")
print(ma_pd[window_size: window_size + 10])

print(f"\nPandas Execution Time: {end - start:.6f} seconds")
pd_time = end - start

# -----------------------------------
# Verify Numerical Equality
# -----------------------------------

print(
    "\nOutputs Match:",
    np.allclose(ma_cpp, ma_pd)
)


print(f"\nSpeedup: {pd_time / cpp_time:.2f}x")
```
